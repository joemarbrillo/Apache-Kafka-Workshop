```python
base = make_order(9001, random.Random(7))
no_city = {k:v for k, v in base.items() if k != "city"}
zero_qty = dict(base, order_id="o-9002", qty=0)
bad = [b"this is not json", json.dumps(no_city).encode(), json.dumps(zero_qty).encode()]
```

```python
producer = Producer({"bootstrap.servers": "localhost:9092"})
```

```python
for value in bad:
    producer.produce("orders", key="c-bad", value=value)

producer.flush()
print(f"send {len(bad)} broken orders")
```
