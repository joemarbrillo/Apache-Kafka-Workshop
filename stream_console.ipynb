```python
spark = get_spark("stream-console")
```

```python
try:
  orders = read_orders(spark)
except Exception as e:
  cause = e.java_exception
  while cause is not None:
    print(cause.toString())
    cause = cause.getCause()
```

```python
query = (orders.select("partition", "offset", "order_id", "city", "qty", "revenue", "is_valid")
         .writeStream.format("console")
         .option("numRows", 8)
         .trigger(availableNow=True)
         .start()
         )
query.awaitTermination()
print("rows read in the first batch", query.recentProgress[0]["numInputRows"])
```
